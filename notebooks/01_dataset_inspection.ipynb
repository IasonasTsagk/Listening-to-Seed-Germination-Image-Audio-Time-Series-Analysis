# Phase 1 — Dataset inspection and global image change

This notebook reads raw images without modifying them. Its global change signal is exploratory and is **not** a germination label.

```python
from pathlib import Path
import sys

current_directory = Path.cwd().resolve()
candidates = [current_directory, *current_directory.parents, current_directory / 'seed-germination']
PROJECT_ROOT = next((path for path in candidates if (path / 'src').is_dir()), None)
if PROJECT_ROOT is None:
    raise RuntimeError('Could not find the project root containing src/. Start Jupyter in or above seed-germination/.')
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import ChangeConfig, ProjectPaths
from src.io_utils import build_inventory, inventory_summary, save_csv
from src.timeseries import global_change_timeseries, plot_global_change

paths = ProjectPaths(root=PROJECT_ROOT)
paths.raw_images
```

```python
inventory = build_inventory(paths.raw_images)
summary = inventory_summary(inventory)
save_csv(inventory, paths.tables / 'dataset_inventory.csv')
save_csv(__import__('pandas').DataFrame([summary]), paths.tables / 'dataset_summary.csv')
summary
```

```python
inventory
```

```python
change_config = ChangeConfig()  # Document any threshold changes before interpreting them.
global_change = global_change_timeseries(
    inventory, change_config, checkpoint_path=paths.features / 'global_image_change.csv'
)
global_change.head()
```

```python
if global_change.empty:
    print('No consecutive readable image pairs yet. Add images to data/raw/images/.')
else:
    plot_global_change(global_change, paths.figures / 'global_image_change.png')
```
